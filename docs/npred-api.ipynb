```python
from gammapy.datasets import MapDataset
from gammapy.modeling.models import Models, SkyModel, FoVBackgroundModel
from astropy.coordinates import SkyCoord
from gammapy.modeling.models.npred import NPredModel, NPredModels
```

```python
dataset_cta = MapDataset.read(
    "$GAMMAPY_DATA/cta-1dc-gc/cta-1dc-gc.fits.gz", name="dataset-cta"
)

print(dataset_cta)
```

```python
model = SkyModel.create("pl", "point", name="gc")
model.spatial_model.position = SkyCoord("0d", "0d", frame="galactic")
model_bkg = FoVBackgroundModel(dataset_name="dataset-cta")

models = Models([model, model_bkg])
```

```python
npred_models = NPredModels.from_map_dataset(models=models, dataset=dataset_cta)
```

```python
%%time
npred = npred_models.npred(parameters=models.parameters, dataset=dataset_cta)
```
